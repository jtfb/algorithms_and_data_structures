
-   Copy the solution your partner wrote. 


```python
# Your answer here

#Input your list here: 
nums = [2,3,5,7]

def missing_num(nums) -> int:
    n = max(nums)
    num_set = set(nums)
    full_set = set(range(n + 1))
    missing_numbers = list(full_set - num_set)
    missing_numbers_sorted  =  sorted(missing_numbers)
    
    if missing_numbers_sorted:
        return missing_numbers_sorted
    else:
        return [-1]
    
missing_num(nums)
```


-   Explain why their solution works in your own words.


```python
# Your answer here
# It works because it sets a difference operation to find all the elements that are missing from <nums> and at the end it sorts that out to ensure everything is in ascending order.
```


-   Explain the problem’s time and space complexity in your own words.


```python
# Your answer here

# Time Complexity:
# - finding <max_nums> and creating <num_set> and <full_set> is O(n) as it needs to go through all the elements where n is the length of <nums>
# - Difference operation <full_set - num_set> is O(n + m) since it needs to iterate through both sets where n is <max(nums)>
# - Sorting <missing_numbers_sorted> is O(k log k) where k is the number of <missing_numbers>

# Space Complexity:
# <num_set> is O(n) as it stores all the elements
# <full_set> is O(m) as it stores all the numbers
# <missing_numbers_sorted> is O(m-n)
```
